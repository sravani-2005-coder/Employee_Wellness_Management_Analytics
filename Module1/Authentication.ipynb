```python
import smtplib, re
import random
from email.mime.text import MIMEText

sender = "darsisravani2@gmail.com"
app_password = "owmwdkfjkqilhiau"

# Email Validation Function

def is_valid_email(email):
    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return re.fullmatch(pattern, email) is not None

# Keep Asking Until Valid Email

while True:
    receiver = input("Enter your Email: ").strip()

    if is_valid_email(receiver):
        break
    else:
        print("❌ Invalid Email Address! Please enter a valid email.\n")

otp = random.randint(100000,999999)

msg = MIMEText(f"Your OTP is {otp}")
msg["Subject"] = "MoodMentor OTP"
msg["From"] = sender
msg["To"] = receiver

server = smtplib.SMTP("smtp.gmail.com",587)
server.starttls()
server.login(sender, app_password)
server.sendmail(sender, receiver, msg.as_string())
server.quit()

print("OTP Sent!")
print("OTP =", otp)

entered_otp = input("Enter the OTP received in your email: ")

if entered_otp == str(otp):
    print("✅ Email Verified Successfully!")
else:
    print("❌ Invalid OTP")
```

```python
import jwt
from datetime import datetime, timedelta, UTC

SECRET_KEY = "Sravani123456789012345678901234567890"
# For HS256, the recommended secret key length is at least 32 bytes.

payload = {
    "email": receiver,
    "exp": datetime.now(UTC) + timedelta(hours=1)
}

token = jwt.encode(payload, SECRET_KEY, algorithm="HS256")

print("JWT Token:")
print(token)
```

```python
try:
    decoded = jwt.decode(token, SECRET_KEY, algorithms=["HS256"])
    print("Token is Valid")
    print(decoded)
except jwt.ExpiredSignatureError:
    print("Token Expired")
except jwt.InvalidTokenError:
    print("Invalid Token")
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd

# --------------------------------------------------
# Page Config
# --------------------------------------------------
st.set_page_config(
    page_title="MoodMentor",
    page_icon="🌸",
    layout="centered"
)

# --------------------------------------------------
# Custom CSS
# --------------------------------------------------
st.markdown("""
<style>

/* Background */
.stApp{
    background: linear-gradient(180deg,#FFFDFB,#FFF5F8);
    color:#444444;
    font-family: 'Segoe UI', sans-serif;
}

/* Sidebar */
section[data-testid="stSidebar"]{
    background:#FCEEF4;
    border-right:2px solid #F6D8E4;
}

/* Titles */
h1{
    color:#C76B98;
    text-align:center;
    font-weight:bold;
}

h2,h3{
    color:#9B5C82;
}

/* Buttons */
.stButton>button{
    background:#F6CFE0;
    color:#6B4860;
    border:none;
    border-radius:12px;
    padding:0.6rem 1.4rem;
    font-weight:600;
}

.stButton>button:hover{
    background:#EFB6CF;
    color:white;
}

/* Inputs */
.stTextInput input{
    border-radius:12px;
    border:1px solid #F3C9D9;
    background:#FFFFFF;
}

/* Metrics */
div[data-testid="metric-container"]{
    background:#FFFFFF;
    border:1px solid #F4DCE7;
    padding:15px;
    border-radius:16px;
    box-shadow:0 3px 8px rgba(0,0,0,0.05);
}

/* Alerts */
div[data-testid="stAlert"]{
    border-radius:12px;
}

/* Main container */
.block-container{
    padding-top:2rem;
    padding-bottom:2rem;
}

</style>
""", unsafe_allow_html=True)

# --------------------------------------------------
# Session State
# --------------------------------------------------
if "logged_in" not in st.session_state:
    st.session_state.logged_in = False
if "uploaded_df" not in st.session_state:
    st.session_state.uploaded_df = None
# --------------------------------------------------
# Sidebar
# --------------------------------------------------
if not st.session_state.logged_in:

    choice = st.sidebar.selectbox(
        "🌸 Menu",
        ["Home", "Register", "Login"]
    )

else:

    choice = st.sidebar.radio(
        "🌸 Navigation",
        [
            "Dashboard",
            "Analyze",
            "History",
            "Profile",
            "Logout"
        ]
    )

# --------------------------------------------------
# HOME
# --------------------------------------------------
if choice == "Home":

    st.title("🌸 MoodMentor")

    st.subheader("AI-Powered Emotional Wellness System")

    st.markdown("---")

    st.markdown("""
### Welcome 💜

MoodMentor helps you understand your emotions with AI.

✨ Track your mood

💖 Analyze your emotions

📊 View your history

🌱 Improve your emotional well-being
""")

# --------------------------------------------------
# REGISTER
# --------------------------------------------------
elif choice == "Register":

    st.title("🌸 Create Account")

    email = st.text_input("📧 Email")

    password = st.text_input(
        "🔒 Password",
        type="password"
    )

    confirm = st.text_input(
        "🔐 Confirm Password",
        type="password"
    )

    st.markdown("")

    if st.button("📨 Send OTP"):
        st.success("OTP Sent Successfully!")

    otp = st.text_input("Enter OTP")

    if st.button("✅ Verify OTP"):
        st.success("Email Verified Successfully!")

# --------------------------------------------------
# LOGIN
# --------------------------------------------------
elif choice == "Login":

    st.title("🔐 Login")

    login_email = st.text_input("📧 Email")

    login_password = st.text_input(
        "Password",
        type="password"
    )

    if st.button("Login"):

        if login_email and login_password:

            st.success("Login Successful!")

            st.session_state.logged_in = True

            st.rerun()

        else:

            st.error("Please enter Email and Password.")

# --------------------------------------------------
# DASHBOARD
# --------------------------------------------------
elif choice == "Dashboard":

    st.title("😊 Dashboard")

    st.success("Welcome back!")

    col1, col2 = st.columns(2)

    with col1:
        st.metric("😊 Current Mood", "Happy")

    with col2:
        st.metric("💜 Sentiment", "Positive")

    st.markdown("---")

    st.info("Have a wonderful day! 🌸")

# --------------------------------------------------
# ANALYZE
# --------------------------------------------------
elif choice == "Analyze":

    st.title("🧠 Mood Analysis")

    uploaded_file = st.file_uploader(
        "📁 Upload CSV File",
        type=["csv"]
    )

    if uploaded_file is not None:

        df = pd.read_csv(uploaded_file)

        st.session_state.uploaded_df = df

        st.success("✅ CSV Uploaded Successfully!")

        col1, col2 = st.columns(2)

        with col1:
            st.metric("Rows", df.shape[0])

        with col2:
            st.metric("Columns", df.shape[1])

        st.subheader("Preview")

        st.dataframe(df.head())

        if st.button("Analyze CSV"):

            st.success("Mood analysis completed successfully!")

            st.write(df.head())

# --------------------------------------------------
# HISTORY
# --------------------------------------------------
elif choice == "History":

    st.title("📊 Mood History")

    st.write("No mood records available yet.")

# --------------------------------------------------
# PROFILE
# --------------------------------------------------
elif choice == "Profile":

    st.title("👤 Profile")

    st.write("Name : User")

    st.write("Email : user@example.com")

# --------------------------------------------------
# LOGOUT
# --------------------------------------------------
elif choice == "Logout":

    st.session_state.logged_in = False

    st.success("Logged out successfully!")

    st.rerun()
```
